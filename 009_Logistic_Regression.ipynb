# Logistic Regression
This notebook demonstrates how to build, evaluate, and interpret a **Logistic Regression** model for binary classification using scikit-learn.
Logistic Regression predicts the probability of a binary event (0 or 1, No or Yes) using the standard logistic (sigmoid) function:
$$P(y=1|X) = \frac{1}{1 + e^{-(\beta_0 + \beta_1 X)}}$$

Key topics covered:
1. Creating synthetic training mileage and race completion data.
2. Encoding categorical binary outcomes using `OrdinalEncoder`.
3. Visualizing feature-target distributions and class balance.
4. Splitting the dataset into train and test sets.
5. Fitting a `LogisticRegression` classifier.
6. Evaluating predictions using accuracy, confusion matrix, and classification report.

```python
df = pd.DataFrame(data=d)
```

### Encode Target Variable
Use `OrdinalEncoder` to convert categorical labels (`'no'`, `'yes'`) into numerical values ($0$ and $1$).

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
finished_race = ['no', 'yes']
```

```python
enc = OrdinalEncoder(categories=[finished_race])
```

```python
df['completed_50m_ultra'] = enc.fit_transform(df[['completed_50m_ultra']])
```

```python
df
```

### Exploratory Data Analysis & Visualization
Visualize the relationship between weekly miles and race completion probability via scatter plot and check class distribution with a count plot.

```python
from matplotlib import pyplot as plt
```

```python
plt.scatter(df['miles_per_week'], df['completed_50m_ultra'])
```

```python
import seaborn as sns
```

```python
sns.countplot(x='completed_50m_ultra', data=df)
```

### Define Features and Target
Separate the feature `X` (`miles_per_week`) and target `y` (`completed_50m_ultra`).

```python
X = df.iloc[:, 0:1]
y = df.iloc[:, 1]
```

### Train-Test Split
Split data into an 80% training set and a 20% test set with fixed `random_state=11`.

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=11 )
```

### Train Logistic Regression Model
Instantiate `LogisticRegression` and fit it on the training data.

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

### Make Predictions & Evaluate Accuracy
Generate binary predictions for the test set and calculate mean accuracy score.

```python
y_pred = model.predict(X_test)
```

```python
model.score(X_test, y_test)
```

### Detailed Model Evaluation: Confusion Matrix & Classification Report
Compute:
- **Confusion Matrix**: Identifies true negatives, false positives, false negatives, and true positives.
- **Classification Report**: Summarizes Precision, Recall, F1-Score, and Support for each class.

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```
