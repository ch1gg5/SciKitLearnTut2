# Scaling
This notebook demonstrates how to scale numerical features using `StandardScaler` and `MinMaxScaler`.
Scaling is crucial because many algorithms (such as KNN, SVM, Logistic Regression, and Neural Networks) are sensitive to the scale of input features. Standardizing or normalizing features helps models converge faster and prevents features with large magnitudes from dominating distance calculations.

- **`StandardScaler`**: Standardizes features by removing the mean and scaling to unit variance ($z = \frac{x - \mu}{\sigma}$).
- **`MinMaxScaler`**: Transforms features by scaling each feature to a specified range (default $[0, 1]$: $x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}$).

```python
import pandas as pd
```

### Load and prepare data
Load the baseball dataset and drop non-numerical/irrelevant columns (`PLAYER` and `CS`).

```python
df = pd.read_csv('500hits.csv', encoding='latin-1')
```

```python
df = df.drop(columns=['PLAYER', 'CS'])
```

```python
df.head()
```

### Select feature subsets for comparison
Create two identical copies of the numerical features (`X1` and `X2`) to demonstrate the effects of both scalers.

```python
X1 = df.iloc[:, 0:13]
X2 = df.iloc[:, 0:13]
```

### 1. StandardScaler (Z-Score Standardization)
Standardizes features such that they have a mean ($\mu$) of 0 and a standard deviation ($\sigma$) of 1.

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaleStandard = StandardScaler()
```

```python
X1 = scaleStandard.fit_transform(X1)
X1 = pd.DataFrame(X1, columns=df.columns[0:13])
```

### Inspect StandardScaler Results
Check the transformed feature values and summary statistics. Note that means are centered around 0 and standard deviations equal 1.

```python
X1.head()
```

```python
X1.describe().round(3)
```

### 2. MinMaxScaler (Normalization to [0, 1])
Scales each feature independently such that the minimum value is 0 and the maximum value is 1.

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaleMinMax = MinMaxScaler(feature_range=(0,1))
```

```python
X2 = scaleMinMax.fit_transform(X2)
X2 = pd.DataFrame(X2, columns=df.columns[0:13])
```

### Inspect MinMaxScaler Results
Check the transformed feature values and summary statistics. Note that the minimum is 0 and maximum is 1 for all columns.

```python
X2.head()
```

```python
X2.describe().round(3)
```
