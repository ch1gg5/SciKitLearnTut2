# Decision Tree Classifier
This notebook demonstrates how to train, evaluate, and tune a **Decision Tree Classifier** using scikit-learn.
Decision Trees are non-parametric supervised learning models that predict target values by learning simple decision rules inferred from data features.

Key topics covered:
1. Loading and preprocessing the baseball Hall of Fame (`HOF`) dataset.
2. Splitting data into training and test sets.
3. Building a baseline Decision Tree with default parameters.
4. Evaluating performance using Confusion Matrix, Precision, Recall, and F1-score.
5. Interpreting feature importances (Gini importance).
6. Post-pruning and hyperparameter tuning using Information Gain (`criterion='entropy'`) and Cost-Complexity Pruning (`ccp_alpha`).

```python
import pandas as pd
```

### Load the Dataset
Load player statistics from `500hits.csv`.

```python
df = pd.read_csv('500hits.csv', encoding='latin-1')
```

```python
df.head()
```

### Data Cleaning and Feature Selection
Drop non-numerical/irrelevant identifier columns (`PLAYER` and `CS`).
Separate the feature matrix `X` (first 13 columns containing player statistics) and the target variable `y` (`HOF` indicator: 1 for Hall of Fame, 0 otherwise).

```python
df = df.drop(columns= ['PLAYER', 'CS'])
```

```python
X = df.iloc[:, 0:13]
```

```python
y = df.iloc[:, 13]
```

### Train-Test Split
Split the dataset into 80% training data and 20% test data to evaluate generalization performance.

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y,random_state=17, test_size=0.2)
```

### Initialize Decision Tree Classifier
Instantiate the default `DecisionTreeClassifier` (using Gini impurity as split criterion) and inspect its default hyperparameters.

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtc = DecisionTreeClassifier()
```

```python
dtc.get_params()
```

### Train the Model & Generate Predictions
Fit the decision tree on the training dataset and predict outcomes for the unseen test set.

```python
dtc.fit(X_train, y_train)
```

```python
y_pred = dtc.predict(X_test)
```

### Model Evaluation: Confusion Matrix & Classification Report
Evaluate classification performance:
- **Confusion Matrix**: Shows true positives, true negatives, false positives, and false negatives.
- **Classification Report**: Shows precision, recall, and F1-score for each class.

```python
from sklearn.metrics import confusion_matrix
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

### Feature Importance Analysis
Inspect feature importances computed by the decision tree. Higher values indicate features that contributed more significantly to reducing impurity across splits.

```python
dtc.feature_importances_
```

```python
features = pd.DataFrame(dtc.feature_importances_, index= X.columns)
```

```python
features.head(15)
```

### Hyperparameter Tuning & Pruning
Train a second decision tree with:
- `criterion='entropy'`: Uses Information Gain (Shannon entropy) instead of Gini impurity.
- `ccp_alpha=0.04`: Cost-Complexity Pruning parameter to prune weak branches and prevent overfitting.

```python
dtc2 = DecisionTreeClassifier(criterion= 'entropy', ccp_alpha=0.04)
```

```python
dtc2.fit(X_train, y_train)
```

### Evaluate Pruned Model & Compare Feature Importances
Evaluate the pruned decision tree and examine how pruning impacted the feature importance distribution.

```python
y_pred2 = dtc2.predict(X_test)
```

```python
print(confusion_matrix(y_test, y_pred2))
```

```python
print(classification_report(y_test, y_pred2))
```
