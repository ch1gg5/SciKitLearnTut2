# One Hot Encoder
This notebook demonstrates how to use `OneHotEncoder` to convert categorical variables into a machine-readable format.
One-hot encoding creates a binary column for each category, which is useful when there is no ordinal relationship between the categories.

```python
import pandas as pd
```

### Load sample data
We create a sample DataFrame with sales, city, and size information.

```python
d = {'sales': [100000,222000,1000000,522000,111111,222222,1111111,20000,75000,90000,1000000,10000],
'city': ['Tampa','Tampa','Orlando','Jacksonville','Miami','Jacksonville','Miami','Miami','Orlando','Orlando','Orlando','Orlando'],
'size': ['Small', 'Medium','Large','Large','Small','Medium','Large','Small','Medium','Medium','Medium','Small',]}
```

```python
df = pd.DataFrame(data=d)
```

```python
df.head()
```

### Prepare for encoding
We check the unique cities and initialize the `OneHotEncoder`.

```python
df['city'].unique()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False).set_output(transform="pandas")
```

### Transform and merge
We transform the 'city' column and concatenate the resulting one-hot encoded columns back to the DataFrame.

```python
ohetransform = ohe.fit_transform(df[['city']])
```

```python
ohetransform.head()
```

```python
df = pd.concat([df, ohetransform], axis=1).drop(columns=['city'])
```
