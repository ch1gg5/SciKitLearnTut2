# # Test Train Split
This notebook demonstrates how to split a dataset into training and testing sets.
This is a critical step in machine learning to evaluate how well a model generalizes to unseen data.
We use `train_test_split` to divide the data, ensuring the model's performance can be assessed independently of the training data.


```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

### Load the data
We load the dataset which contains player statistics.

```python
df = pd.read_csv('500hits.csv', encoding='latin-1')
```

```python
df.head()
```

### Prepare features (X) and target (y)
We drop non-feature columns ('PLAYER') and the target variable ('HOF') from X, and assign 'HOF' to y.

```python
X = df.drop(columns=['PLAYER', 'HOF'])
y = df['HOF']
```

```python
X.head()
```

### Split the data
We perform the split. `test_size=0.2` means 20% of the data will be for testing. `random_state` ensures the split is reproducible.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=11, test_size=0.2)
```
