# K-Nearest Neighbors (KNN) Classifier
This notebook demonstrates how to build, scale, and evaluate a **K-Nearest Neighbors (KNN) Classifier** using scikit-learn.
KNN is an instance-based (lazy) supervised learning algorithm that classifies a new sample based on the majority class of its $k$ closest data points in the feature space (using distance metrics like Euclidean distance).

Key topics covered:
1. Loading and preprocessing baseball player data from `500hits.csv`.
2. Splitting data into training and test sets.
3. Feature scaling using `MinMaxScaler` (critical for distance-based algorithms like KNN).
4. Training a `KNeighborsClassifier` with $k=8$.
5. Evaluating model performance using accuracy score, confusion matrix, and classification report.
6. Inspecting the number of fitted samples.

```python
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.metrics import confusion_matrix, classification_report
```

```python
# code out the model
```

### Load Dataset
Read player statistics from `500hits.csv`.

```python
df = pd.read_csv('500hits.csv', encoding='latin-1')
```

```python
df.head()
```

### Data Cleaning and Feature Preparation
Drop non-numerical/irrelevant identifier columns (`PLAYER` and `CS`).
Separate the feature set `X` (all batting statistics) and target `y` (`HOF`).

```python
df = df.drop(columns=['PLAYER', 'CS'])
```

```python
df.head()
```

```python
X = df.drop(columns=['HOF'])
```

```python
y = df['HOF']
```

### Train-Test Split
Split data into an 80% training set and a 20% test set with fixed `random_state=11`.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=11, test_size=0.2)
```

### Feature Scaling (MinMaxScaler)
Because KNN calculates geometric distances between points, features with large numeric scales would dominate the distance calculation without normalization.
We fit the `MinMaxScaler` on `X_train` and transform both `X_train` and `X_test` into the range $[0, 1]$.

```python
scaler = MinMaxScaler(feature_range=(0, 1))
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

### Initialize and Train KNN Model
Instantiate `KNeighborsClassifier` with $k=8$ neighbors and fit it on the scaled training data.

```python
knn = KNeighborsClassifier(n_neighbors=8)
```

```python
knn.fit(X_train, y_train)
```

### Make Predictions & Evaluate Model
Generate predictions on the test set and calculate accuracy.

```python
y_pred = knn.predict(X_test)
```

```python
knn.score(X_test, y_test)
```

### Detailed Evaluation: Confusion Matrix & Classification Report
Compute:
- **Confusion Matrix**: Overview of correct and incorrect classifications.
- **Classification Report**: Precision, Recall, and F1-scores.

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

### Inspect Fitted Samples
Check the total number of training instances stored in the KNN model.

```python
print(knn.n_samples_fit_)
```
