# Ordinal Encoder
This notebook demonstrates how to use `OrdinalEncoder` to convert categorical variables with an inherent order (ordinal) into numerical values.
Unlike one-hot encoding, this preserves the order information, which is important for variables like size (Small, Medium, Large).

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
```

### Load sample data
We create a sample DataFrame with sales, city, and size information.

```python
d = {'sales': [100000,222000,1000000,522000,111111,222222,1111111,20000,75000,90000,1000000,10000],
'city': ['Tampa','Tampa','Orlando','Jacksonville','Miami','Jacksonville','Miami','Miami','Orlando','Orlando','Orlando','Orlando'],
'size': ['Small', 'Medium','Large','Large','Small','Medium','Large','Small','Medium','Medium','Medium','Small',]}
```

```python
df = pd.DataFrame(data=d)
df.head()
```

### Set ordering and transform
We define the order for the 'size' category ('Small' < 'Medium' < 'Large') and use `OrdinalEncoder` to transform the data.

```python
sizes = ['Small', 'Medium', 'Large']
```

```python
enc = OrdinalEncoder(categories=[sizes])
enc.fit_transform(df[['size']].head())
```

```python
df['size'] = enc.fit_transform(df[['size']])
```
